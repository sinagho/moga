```python
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.functional as F
import torch.optim as optim
```

```python
from lhunet.modules.cnn.attentions import DLKA3D_Block_onTensor
from lhunet.models.v12 import LHUNet
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
block = DLKA3D_Block_onTensor(64).to(device)
block.train()

criteria = nn.MSELoss()
optimizer = optim.Adam(block.parameters(), lr=0.001)

x = torch.randn(1, 64, 32, 32, 32).to(device)
y = torch.randn(1, 64, 32, 32, 32).to(device)


optimizer.zero_grad()
outputs = block(x)
print(outputs.size())

loss = criteria(outputs, y)
print(loss.item())
loss.backward()
```

```python
scaler = torch.cuda.amp.GradScaler()

with torch.cuda.amp.autocast():
    optimizer.zero_grad()
    outputs = block(x)
    loss = criteria(outputs, y)
    loss.backward()
```
